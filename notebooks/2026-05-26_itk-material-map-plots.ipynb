```python
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("material-maps-ATLAS-P2-RUN4-04-00-00-ITk-HGTD.json") as f:
    material = json.load(f)
```

```python
material.keys();
material["Surfaces"].keys();
material["Surfaces"]["entries"][0];
```

```python
def extractMaterialAxes(data):
    axes = [bd for bd in data["binningdata"]]
    return axes

def extractMaterialMatrix(data, key=None, index=None):
    def extractCell(cell):
        if key is not None:
            cell = cell[key]
        if cell is None:
            return float("nan")
        if index is not None:
            cell = cell[index]
        return cell
    matrix = []
    for row in data:
        new_row = []
        for cell in row:
            new_cell = extractCell(cell)
            new_row.append(new_cell)
        matrix.append(new_row)
    return np.array(matrix)

# eRadiationLength = 0,
# eInteractionLength = 1,
# eRelativeAtomicMass = 2,
# eNuclearCharge = 3,
# eMolarDensity = 4,
```

```python
entry = 6

axes = extractMaterialAxes(material["Surfaces"]["entries"][entry]["value"]["material"]["binUtility"])
map = extractMaterialMatrix(material["Surfaces"]["entries"][entry]["value"]["material"]["data"], key="material", index=0)

plt.imshow(map, origin='lower')
plt.colorbar(orientation='vertical')

plt.xlabel(axes[0]["value"])
plt.ylabel(axes[1]["value"])
```
